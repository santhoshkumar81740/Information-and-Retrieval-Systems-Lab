```python
import re
from collections import defaultdict

documents = {
    "D1": "Artificial Intelligence enables machines to perform tasks that normally require human intelligence such as learning reasoning and decision making.",

    "D2": "Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data and improve performance without explicit programming.",

    "D3": "Python is a popular programming language used for Artificial Intelligence Machine Learning Data Science and Web Development.",

    "D4": "Information Retrieval is the process of obtaining relevant information from a collection of documents using search techniques and indexing methods.",

    "D5": "Natural Language Processing helps computers understand analyze and generate human language for applications such as chatbots and translation.",

    "D6": "Data Mining discovers useful patterns and knowledge from large datasets using statistical methods machine learning and database systems.",

    "D7": "Cloud Computing provides storage servers databases networking and software services over the internet making computing resources easily accessible.",

    "D8": "Cybersecurity protects computers networks and digital information from malware phishing ransomware hackers and unauthorized access.",

    "D9": "Big Data refers to extremely large datasets that require modern technologies such as Hadoop Spark and distributed computing for processing.",

    "D10": "The Internet of Things connects smart devices sensors and machines through the internet allowing communication automation and remote monitoring."
}

index = defaultdict(set)

for doc, text in documents.items():
    for word in re.findall(r'\b[a-zA-Z]+\b', text.lower()):
        index[word].add(doc)

print("Vocabulary Size:", len(index))

for word in sorted(index):
    print(word, ":", sorted(index[word]))
```
