```python
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Load the standard dataset
categories = ['sci.space', 'rec.sport.baseball']

data = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

documents = data.data
print("Total Documents:", len(documents))

# Step 2: Convert documents into TF-IDF matrix
vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=1000
)

X = vectorizer.fit_transform(documents)
print("TF-IDF Matrix Shape:", X.shape)

# Step 3: Matrix decomposition using SVD
svd = TruncatedSVD(n_components=2, random_state=42)
X_lsi = svd.fit_transform(X)

print("\nLSI Matrix Shape:", X_lsi.shape)
print("\nFirst Five LSI Vectors:")
print(X_lsi[:5])

# Step 4: Display explained variance
print("\nExplained Variance Ratio:")
print(svd.explained_variance_ratio_)

# Step 5: Select a query document
query_index = 0
query_vector = X_lsi[query_index].reshape(1, -1)

# Step 6: Calculate similarity with all documents
similarities = cosine_similarity(query_vector, X_lsi)[0]

# Step 7: Rank documents by similarity
ranked_indices = np.argsort(similarities)[::-1]

print("\nQuery Document:")
print(documents[query_index][:300])

print("\nRanked Documents:")
rank = 1

for index in ranked_indices:
    if index == query_index:
        continue

    print("\nRank:", rank)
    print("Document Index:", index)
    print("Similarity Score:", round(similarities[index], 4))
    print("Document:", documents[index][:200].replace("\n", " "))

    rank += 1

    if rank > 10:
        break

# Step 8: Display top terms in each latent component
terms = vectorizer.get_feature_names_out()

for i, component in enumerate(svd.components_):
    indices = np.argsort(component)[-10:][::-1]
    print("\nTop Terms in Component", i + 1)
    print([terms[j] for j in indices])

print("\nMatrix Decomposition and LSI completed successfully!")
```
