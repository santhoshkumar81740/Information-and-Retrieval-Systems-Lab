```python
import re
from collections import defaultdict

documents = {
    "D1": "Python is a programming language.",
    "D2": "Machine learning uses Python.",
    "D3": "Information retrieval uses an inverted index."
}

index = defaultdict(set)

for doc, text in documents.items():
    for word in re.findall(r'\b[a-zA-Z]+\b', text.lower()):
        index[word].add(doc)

print("Vocabulary Size:", len(index))

for word in sorted(index):
    print(word, ":", sorted(index[word]))
```
