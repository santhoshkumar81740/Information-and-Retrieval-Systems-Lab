```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import quote

def search_news(topic, websites=None):
    # Create search query
    query = topic

    # Restrict search to selected websites
    if websites:
        site_query = " OR ".join("site:" + site for site in websites)
        query = topic + " " + site_query

    # Bing search URL
    url = "https://www.google.com/search?q=" + quote(query)

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print("Unable to access search engine.")
        return

    soup = BeautifulSoup(response.text, "html.parser")

    print("\nNews Stories Related To:", topic)
    print("-" * 60)

    # Extract search results
    results = soup.select("div.MjjYud")

    count = 0

    for result in results:
        title = result.select_one("h3")
        link = result.select_one("a")

        if title and link:
            print("\nTitle:", title.get_text())
            print("URL:", link.get("href"))

            count += 1

            if count == 10:
                break

    if count == 0:
        print("No news stories found.")


# Main program
topic = input("Enter the topic to search: ")

choice = input(
    "Do you want to restrict the search to selected websites? (yes/no): "
)

websites = []

if choice.lower() == "yes":
    sites = input(
        "Enter websites separated by commas (example: bbc.com, reuters.com): "
    )

    websites = [site.strip() for site in sites.split(",")]

search_news(topic, websites)
```
